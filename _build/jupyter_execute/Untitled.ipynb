Istilah yang wajib hafal
Dari PDF ini, ada beberapa istilah yang bakal terus muncul di mata kuliah Struktur Data:
Node — kotak yang menyimpan data. Kalau kamu lihat diagram tree, setiap kotak itu adalah node.
Edge — garis penghubung antar node. Satu node hanya bisa punya satu edge masuk, tapi bisa punya banyak edge keluar.
Root — node paling atas. Tidak ada edge yang masuk ke sini. Ini titik awal semua traversal.
Path — urutan node dari root sampai tujuan. 
Parent & Children — hubungan langsung. Node C: adalah parent dari Program Files, Users, Windows. Ketiga itu children-nya.
Siblings — node-node yang parent-nya sama.
Leaf node — node yang tidak punya children sama sekali. Ujung jalan buntu di tree.
Level — Root itu level 0. Semakin dalam, levelnya makin besar.
Height — level tertinggi yang ada di tree itu.

    Binary Tree — Tree dengan Aturan Ketat
Binary Tree adalah Tree yang setiap node-nya maksimal punya dua children: kiri dan kanan. Tidak lebih.
Implementasinya di Python pakai class, karena setiap node perlu menyimpan tiga hal: nilai (key), pointer ke child kiri, dan pointer ke child kanan.    

```python
class BinaryTree:
    def __init__(self, root):
        self.key = root
        self.leftChild = None
        self.rightChild = None
```
