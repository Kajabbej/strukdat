```python
# --- BINARY SEARCH TREE (INSERT & SEARCH) ---
class BSTNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        if self.root is None:
            self.root = BSTNode(value)
        else:
            self._insert_recursive(self.root, value)

    def _insert_recursive(self, curr_node, value):
        if value < curr_node.value:
            if curr_node.left is None:
                curr_node.left = BSTNode(value)
            else:
                self._insert_recursive(curr_node.left, value)
        elif value > curr_node.value:
            if curr_node.right is None:
                curr_node.right = BSTNode(value)
            else:
                self._insert_recursive(curr_node.right, value)

    def search_recursive(self, value):
        return self._search_recursive(self.root, value)

    def _search_recursive(self, curr_node, value):
        if curr_node is None:
            return False
        if curr_node.value == value:
            return True
        elif value < curr_node.value:
            return self._search_recursive(curr_node.left, value)
        else:
            return self._search_recursive(curr_node.right, value)

    def search_iterative(self, value):
        curr = self.root
        while curr is not None:
            if curr.value == value:
                return True
            elif value < curr.value:
                curr = curr.left
            else:
                curr = curr.right
        return False
```

```python
# --- BST MIN, MAX, & DELETE OPERATIONS ---
class AdvancedBinarySearchTree(BinarySearchTree):
    def find_min(self):
        if self.root is None:
            return None
        return self._find_min_node(self.root).value

    def _find_min_node(self, node):
        curr = node
        while curr.left is not None:
            curr = curr.left
        return curr

    def find_max(self):
        if self.root is None:
            return None
        curr = self.root
        while curr.right is not None:
            curr = curr.right
        return curr.value

    def delete(self, value):
        self.root = self._delete_recursive(self.root, value)

    def _delete_recursive(self, curr_node, value):
        if curr_node is None:
            return curr_node

        if value < curr_node.value:
            curr_node.left = self._delete_recursive(curr_node.left, value)
        elif value > curr_node.value:
            curr_node.right = self._delete_recursive(curr_node.right, value)
        else:
            # Kasus 1: Node tanpa anak
            if curr_node.left is None and curr_node.right is None:
                return None
            
            # Kasus 2: Node dengan 1 anak
            if curr_node.left is None:
                return curr_node.right
            elif curr_node.right is None:
                return curr_node.left

            # Kasus 3: Node dengan 2 anak (ambil penerus inorder/terkecil dari subtree kanan)
            temp = self._find_min_node(curr_node.right)
            curr_node.value = temp.value
            curr_node.right = self._delete_recursive(curr_node.right, temp.value)

        return curr_node
```

```python
# --- TRAVERSAL BST (DFS & BFS) ---
class TraversableBST(AdvancedBinarySearchTree):
    def inorder(self):
        res = []
        self._inorder_recursive(self.root, res)
        return res

    def _inorder_recursive(self, node, res):
        if node:
            self._inorder_recursive(node.left, res)
            res.append(node.value)
            self._inorder_recursive(node.right, res)

    def preorder(self):
        res = []
        self._preorder_recursive(self.root, res)
        return res

    def _preorder_recursive(self, node, res):
        if node:
            res.append(node.value)
            self._preorder_recursive(node.left, res)
            self._preorder_recursive(node.right, res)

    def postorder(self):
        res = []
        self._postorder_recursive(self.root, res)
        return res

    def _postorder_recursive(self, node, res):
        if node:
            self._postorder_recursive(node.left, res)
            self._postorder_recursive(node.right, res)
            res.append(node.value)

    def level_order(self):
        # Breadth-First Search (BFS) menggunakan Queue list
        if self.root is None:
            return []
        res = []
        queue = [self.root]
        while len(queue) > 0:
            curr = queue.pop(0)
            res.append(curr.value)
            if curr.left:
                queue.append(curr.left)
            if curr.right:
                queue.append(curr.right)
        return res
```

```python
# --- METODE BANTU DAN VISUALISASI TREE ---
class VisualizableBST(TraversableBST):
    def get_height(self):
        return self._get_height_recursive(self.root)

    def _get_height_recursive(self, node):
        if node is None:
            return -1 # Node kosong memiliki tinggi -1
        left_h = self._get_height_recursive(node.left)
        right_h = self._get_height_recursive(node.right)
        return 1 + max(left_h, right_h)

    def get_leaf_count(self):
        return self._get_leaf_count_recursive(self.root)

    def _get_leaf_count_recursive(self, node):
        if node is None:
            return 0
        if node.left is None and node.right is None:
            return 1
        return self._get_leaf_count_recursive(node.left) + self._get_leaf_count_recursive(node.right)

    def get_node_count(self):
        return self._get_node_count_recursive(self.root)

    def _get_node_count_recursive(self, node):
        if node is None:
            return 0
        return 1 + self._get_node_count_recursive(node.left) + self._get_node_count_recursive(node.right)

    def display(self):
        self._display_recursive(self.root, 0, "Root: ")

    def _display_recursive(self, node, level, prefix):
        if node is not None:
            print(" " * (level * 4) + prefix + str(node.value))
            if node.left is not None or node.right is not None:
                self._display_recursive(node.left, level + 1, "L--- ")
                self._display_recursive(node.right, level + 1, "R--- ")
```

```python
# --- AVL TREE (SELF-BALANCING BST) ---
class AVLNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        self.height = 0

class AVLTree:
    def get_height(self, node):
        if node is None:
            return -1
        return node.height

    def get_balance(self, node):
        if node is None:
            return 0
        return self.get_height(node.left) - self.get_height(node.right)

    def right_rotate(self, y):
        x = y.left
        T2 = x.right
        
        # Rotasi
        x.right = y
        y.left = T2
        
        # Perbarui tinggi node
        y.height = 1 + max(self.get_height(y.left), self.get_height(y.right))
        x.height = 1 + max(self.get_height(x.left), self.get_height(x.right))
        return x

    def left_rotate(self, x):
        y = x.right
        T2 = y.left
        
        # Rotasi
        y.left = x
        x.right = T2
        
        # Perbarui tinggi node
        x.height = 1 + max(self.get_height(x.left), self.get_height(x.right))
        y.height = 1 + max(self.get_height(y.left), self.get_height(y.right))
        return y

    def insert(self, root, value):
        # 1. Lakukan BST Insertion standar
        if root is None:
            return AVLNode(value)
            
        if value < root.value:
            root.left = self.insert(root.left, value)
        elif value > root.value:
            root.right = self.insert(root.right, value)
        else:
            return root # Tidak memperbolehkan duplikasi nilai
            
        # 2. Perbarui tinggi node leluhur
        root.height = 1 + max(self.get_height(root.left), self.get_height(root.right))
        
        # 3. Cari faktor keseimbangan (balance factor)
        balance = self.get_balance(root)
        
        # Jika tidak seimbang, lakukan salah satu dari 4 rotasi berikut:
        # Kasus LL (Left Left)
        if balance > 1 and value < root.left.value:
            return self.right_rotate(root)
            
        # Kasus RR (Right Right)
        if balance < -1 and value > root.right.value:
            return self.left_rotate(root)
            
        # Kasus LR (Left Right)
        if balance > 1 and value > root.left.value:
            root.left = self.left_rotate(root.left)
            return self.right_rotate(root)
            
        # Kasus RL (Right Left)
        if balance < -1 and value < root.right.value:
            root.right = self.right_rotate(root.right)
            return self.left_rotate(root)
            
        return root

    def display(self, node, level=0, prefix="Root: "):
        if node is not None:
            print(" " * (level * 4) + prefix + f"{node.value} (H={node.height})")
            if node.left is not None or node.right is not None:
                self.display(node.left, level + 1, "L--- ")
                self.display(node.right, level + 1, "R--- ")
```

```python
# --- VERIFIKASI BST & AVL TREE SUITE ---
print("=== 1. VERIFIKASI BINARY SEARCH TREE (BST) ===")
bst = VisualizableBST()
data_uji = [50, 30, 70, 20, 40, 60, 80]
for d in data_uji:
    bst.insert(d)
    
print("Visualisasi Struktur BST:")
bst.display()

print("\n--- Traversal BST ---")
print("Inorder   (L-Root-R) :", bst.inorder())
print("Preorder  (Root-L-R) :", bst.preorder())
print("Postorder (L-R-Root) :", bst.postorder())
print("Level-order (BFS)    :", bst.level_order())

print("\n--- Properti BST ---")
print("Tinggi Tree        :", bst.get_height())
print("Jumlah Daun (Leaf) :", bst.get_leaf_count())
print("Total Node         :", bst.get_node_count())
print("Nilai Minimum      :", bst.find_min())
print("Nilai Maksimum      :", bst.find_max())
print("Cari nilai 40      :", bst.search_recursive(40))
print("Cari nilai 99      :", bst.search_iterative(99))

print("\n--- Penghapusan Node ---")
print("Hapus daun (nilai 20):")
bst.delete(20)
bst.display()

print("\nHapus node dengan 1 anak (nilai 70):")
bst.delete(70)
bst.display()

print("\nHapus root (nilai 50):")
bst.delete(50)
bst.display()

print("\n=== 2. VERIFIKASI AVL TREE (SELF-BALANCING) ===")
avl = AVLTree()
avl_root = None
data_avl = [10, 20, 30, 40, 50, 25] # Akan memicu beberapa rotasi penyeimbang

print("Memasukkan data ke AVL secara berurutan...")
for d in data_avl:
    avl_root = avl.insert(avl_root, d)
    
print("Visualisasi AVL Tree (termasuk Tinggi Node):")
avl.display(avl_root)
```
