```python
class Stack:
    def __init__(self, maks=None):
        self.items = []
        self.maks = maks

    def isEmpty(self):
        return len(self.items) == 0

    def isFull(self):
        if self.maks is None:
            return False
        return len(self.items) >= self.maks

    def push(self, item):
        if self.isFull():
            print(f"OVERFLOW! Stack penuh (maksimum {self.maks})")
            return False
        self.items.append(item)
        return True

    def pop(self):
        if self.isEmpty():
            print("UNDERFLOW! Stack kosong")
            return None
        return self.items.pop()

    def peek(self):
        if self.isEmpty():
            return None
        return self.items[-1]

    def size(self):
        return len(self.items)

    def __str__(self):
        return f"Stack: {self.items} (Top: {self.peek()})"
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedStack:
    def __init__(self):
        self.top = None
        self._size = 0

    def isEmpty(self):
        return self.top is None

    def push(self, item):
        new_node = Node(item)
        new_node.next = self.top
        self.top = new_node
        self._size += 1

    def pop(self):
        if self.isEmpty():
            print("UNDERFLOW! LinkedStack kosong")
            return None
        popped = self.top.data
        self.top = self.top.next
        self._size -= 1
        return popped

    def peek(self):
        if self.isEmpty():
            return None
        return self.top.data

    def size(self):
        return self._size

    def __str__(self):
        elements = []
        curr = self.top
        while curr:
            elements.append(str(curr.data))
            curr = curr.next
        return "LinkedStack (Top -> Bottom): " + " -> ".join(elements) if elements else "LinkedStack: Kosong"
```

```python
def check_balanced_parentheses(expression):
    s = Stack()
    pasangan = {')': '(', '}': '{', ']': '['}
    for char in expression:
        if char in "({[":
            s.push(char)
        elif char in ")}]":
            if s.isEmpty() or s.pop() != pasangan[char]:
                return False
    return s.isEmpty()

def decimal_to_binary(dec_num):
    s = Stack()
    if dec_num == 0:
        return "0"
    while dec_num > 0:
        s.push(dec_num % 2)
        dec_num = dec_num // 2
    
    bin_str = ""
    while not s.isEmpty():
        bin_str += str(s.pop())
    return bin_str
```

```python
class Queue:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return len(self.items) == 0

    def enqueue(self, item):
        self.items.insert(0, item)

    def dequeue(self):
        if self.isEmpty():
            print("UNDERFLOW! Queue kosong")
            return None
        return self.items.pop()

    def size(self):
        return len(self.items)

    def __str__(self):
        return f"Queue: {self.items[::-1]} (Depan ke Belakang)"

class CircularQueue:
    def __init__(self, size):
        self.size = size
        self.queue = [None] * size
        self.front = -1
        self.rear = -1

    def isFull(self):
        return (self.rear + 1) % self.size == self.front

    def isEmpty(self):
        return self.front == -1

    def enqueue(self, data):
        if self.isFull():
            print("OVERFLOW! CircularQueue penuh")
            return False
        elif self.isEmpty():
            self.front = 0
            self.rear = 0
            self.queue[self.rear] = data
        else:
            self.rear = (self.rear + 1) % self.size
            self.queue[self.rear] = data
        return True

    def dequeue(self):
        if self.isEmpty():
            print("UNDERFLOW! CircularQueue kosong")
            return None
        temp = self.queue[self.front]
        self.queue[self.front] = None
        if self.front == self.rear:
            self.front = -1
            self.rear = -1
        else:
            self.front = (self.front + 1) % self.size
        return temp

    def __str__(self):
        return f"CircularQueue: {self.queue} | Front Indeks: {self.front}, Rear Indeks: {self.rear}"
```

```python
class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self._size = 0

    def isEmpty(self):
        return self.front is None

    def enqueue(self, item):
        new_node = Node(item)
        if self.rear is None:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self._size += 1

    def dequeue(self):
        if self.isEmpty():
            print("UNDERFLOW! LinkedQueue kosong")
            return None
        temp = self.front
        self.front = self.front.next
        if self.front is None:
            self.rear = None
        self._size -= 1
        return temp.data

    def size(self):
        return self._size

    def __str__(self):
        elements = []
        curr = self.front
        while curr:
            elements.append(str(curr.data))
            curr = curr.next
        return "LinkedQueue (Front -> Rear): " + " -> ".join(elements) if elements else "LinkedQueue: Kosong"
```

```python
class Deque:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return len(self.items) == 0

    def addFront(self, item):
        self.items.append(item)

    def addRear(self, item):
        self.items.insert(0, item)

    def removeFront(self):
        if self.isEmpty():
            print("UNDERFLOW! Deque kosong")
            return None
        return self.items.pop()

    def removeRear(self):
        if self.isEmpty():
            print("UNDERFLOW! Deque kosong")
            return None
        return self.items.pop(0)

    def size(self):
        return len(self.items)

    def __str__(self):
        return f"Deque: {self.items}"

def palindrome_checker(a_string):
    char_deque = Deque()
    cleaned = "".join(a_string.split()).lower()
    
    for char in cleaned:
        char_deque.addFront(char)
        
    still_equal = True
    while char_deque.size() > 1 and still_equal:
        first = char_deque.removeFront()
        last = char_deque.removeRear()
        if first != last:
            still_equal = False
            
    return still_equal
```

```python
def bubble_sort(data, ascending=True):
    arr = data.copy()
    n = len(arr)
    swaps = 0
    comparisons = 0
    for i in range(n - 1):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            condition = arr[j] > arr[j + 1] if ascending else arr[j] < arr[j + 1]
            if condition:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return arr, comparisons, swaps

def selection_sort(data, ascending=True):
    arr = data.copy()
    n = len(arr)
    swaps = 0
    comparisons = 0
    for i in range(n - 1):
        extreme_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            condition = arr[j] < arr[extreme_idx] if ascending else arr[j] > arr[extreme_idx]
            if condition:
                extreme_idx = j
        if extreme_idx != i:
            arr[i], arr[extreme_idx] = arr[extreme_idx], arr[i]
            swaps += 1
    return arr, comparisons, swaps

def insertion_sort(data, ascending=True):
    arr = data.copy()
    n = len(arr)
    comparisons = 0
    shifts = 0
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            condition = arr[j] > key if ascending else arr[j] < key
            if condition:
                arr[j + 1] = arr[j]
                shifts += 1
                j -= 1
            else:
                break
        arr[j + 1] = key
    return arr, comparisons, shifts
```

```python
class HashTableChaining:
    def __init__(self, size=11):
        self.size = size
        self.slots = [[] for _ in range(size)]

    def hash_function(self, key):
        if isinstance(key, str):
            return sum(ord(c) for c in key) % self.size
        return key % self.size

    def put(self, key, value):
        hash_val = self.hash_function(key)
        for i, item in enumerate(self.slots[hash_val]):
            if item[0] == key:
                self.slots[hash_val][i] = (key, value)
                return
        self.slots[hash_val].append((key, value))

    def get(self, key):
        hash_val = self.hash_function(key)
        for item in self.slots[hash_val]:
            if item[0] == key:
                return item[1]
        return None

    def remove(self, key):
        hash_val = self.hash_function(key)
        for i, item in enumerate(self.slots[hash_val]):
            if item[0] == key:
                return self.slots[hash_val].pop(i)
        return None

    def __str__(self):
        result = []
        for idx, bucket in enumerate(self.slots):
            result.append(f"Slot {idx:02d}: {bucket}")
        return "\n".join(result)
```

```python
class HashTableOpenAddressing:
    def __init__(self, capacity=11, prob_type="linear"):
        self.capacity = capacity
        self.prob_type = prob_type  # "linear", "quadratic", "double"
        self.slots = [None] * capacity
        self.data = [None] * capacity
        self.count = 0

    def hash_function(self, key):
        if isinstance(key, str):
            return sum(ord(c) for c in key) % self.capacity
        return key % self.capacity

    def hash_function2(self, key):
        if isinstance(key, str):
            return 7 - (sum(ord(c) for c in key) % 7)
        return 7 - (key % 7)

    def get_load_factor(self):
        return self.count / self.capacity

    def rehash(self):
        old_slots = self.slots
        old_data = self.data
        old_capacity = self.capacity

        self.capacity = old_capacity * 2 + 1
        self.slots = [None] * self.capacity
        self.data = [None] * self.capacity
        self.count = 0
        print(f"\n[REHASHING] Load factor melebihi 0.75. Kapasitas diperbesar dari {old_capacity} ke {self.capacity}")

        for i in range(old_capacity):
            if old_slots[i] is not None and old_slots[i] != "DELETED":
                self.put(old_slots[i], old_data[i])

    def put(self, key, value):
        if self.get_load_factor() >= 0.75:
            self.rehash()

        hash_val = self.hash_function(key)

        if self.slots[hash_val] is None or self.slots[hash_val] == "DELETED" or self.slots[hash_val] == key:
            if self.slots[hash_val] != key:
                self.count += 1
            self.slots[hash_val] = key
            self.data[hash_val] = value
        else:
            step = 1
            next_slot = hash_val
            found = False

            while self.slots[next_slot] is not None and self.slots[next_slot] != "DELETED":
                if self.slots[next_slot] == key:
                    self.data[next_slot] = value
                    found = True
                    break

                if self.prob_type == "linear":
                    next_slot = (hash_val + step) % self.capacity
                elif self.prob_type == "quadratic":
                    next_slot = (hash_val + step ** 2) % self.capacity
                elif self.prob_type == "double":
                    h2 = self.hash_function2(key)
                    next_slot = (hash_val + step * h2) % self.capacity

                step += 1
                if step >= self.capacity:
                    break

            if not found:
                if self.slots[next_slot] is None or self.slots[next_slot] == "DELETED":
                    self.slots[next_slot] = key
                    self.data[next_slot] = value
                    self.count += 1
                else:
                    print(f"ERROR: Gagal memasukkan kunci {key} karena tabel penuh / probe limit!")

    def get(self, key):
        start_slot = self.hash_function(key)
        step = 1
        next_slot = start_slot

        while self.slots[next_slot] is not None:
            if self.slots[next_slot] == key:
                return self.data[next_slot]

            if self.prob_type == "linear":
                next_slot = (start_slot + step) % self.capacity
            elif self.prob_type == "quadratic":
                next_slot = (start_slot + step ** 2) % self.capacity
            elif self.prob_type == "double":
                h2 = self.hash_function2(key)
                next_slot = (start_slot + step * h2) % self.capacity

            step += 1
            if next_slot == start_slot or step >= self.capacity:
                break
        return None

    def remove(self, key):
        start_slot = self.hash_function(key)
        step = 1
        next_slot = start_slot

        while self.slots[next_slot] is not None:
            if self.slots[next_slot] == key:
                val = self.data[next_slot]
                self.slots[next_slot] = "DELETED"
                self.data[next_slot] = None
                self.count -= 1
                return val

            if self.prob_type == "linear":
                next_slot = (start_slot + step) % self.capacity
            elif self.prob_type == "quadratic":
                next_slot = (start_slot + step ** 2) % self.capacity
            elif self.prob_type == "double":
                h2 = self.hash_function2(key)
                next_slot = (start_slot + step * h2) % self.capacity

            step += 1
            if next_slot == start_slot or step >= self.capacity:
                break
        return None

    def __str__(self):
        result = []
        for idx in range(self.capacity):
            result.append(f"Slot {idx:02d}: Key={self.slots[idx]}, Val={self.data[idx]}")
        return "\n".join(result)
```
