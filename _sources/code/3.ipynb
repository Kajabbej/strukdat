# 3. Sorting

```python
# --- BUBBLE SORT ---
def bubble_sort(data, ascending=True):
    arr = data.copy()
    n = len(arr)
    comparisons = 0
    swaps = 0
    
    for i in range(n - 1):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            condition = arr[j] > arr[j + 1] if ascending else arr[j] < arr[j + 1]
            if condition:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1
                swapped = True
        # Optimasi early exit jika tidak ada penukaran
        if not swapped:
            break
            
    return arr, comparisons, swaps
```

```python
# --- SELECTION SORT ---
def selection_sort(data, ascending=True):
    arr = data.copy()
    n = len(arr)
    comparisons = 0
    swaps = 0
    
    for i in range(n - 1):
        extreme_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            condition = arr[j] < arr[extreme_idx] if ascending else arr[j] > arr[extreme_idx]
            if condition:
                extreme_idx = j
                
        if extreme_idx != i:
            arr[i], arr[extreme_idx] = arr[extreme_idx], arr[i]
            swaps += 1
            
    return arr, comparisons, swaps

def selection_sort_names(names, ascending=True):
    arr = names.copy()
    n = len(arr)
    for i in range(n - 1):
        extreme_idx = i
        for j in range(i + 1, n):
            # Menggunakan case-insensitive comparison
            condition = arr[j].lower() < arr[extreme_idx].lower() if ascending else arr[j].lower() > arr[extreme_idx].lower()
            if condition:
                extreme_idx = j
        if extreme_idx != i:
            arr[i], arr[extreme_idx] = arr[extreme_idx], arr[i]
    return arr
```

```python
# --- INSERTION SORT ---
def insertion_sort(data, ascending=True):
    arr = data.copy()
    n = len(arr)
    comparisons = 0
    shifts = 0
    
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            condition = arr[j] > key if ascending else arr[j] < key
            if condition:
                arr[j + 1] = arr[j]
                shifts += 1
                j -= 1
            else:
                break
        arr[j + 1] = key
        
    return arr, comparisons, shifts

def binary_insertion_sort(data, ascending=True):
    # Optimasi mencari posisi sisipan menggunakan Binary Search
    arr = data.copy()
    n = len(arr)
    
    for i in range(1, n):
        key = arr[i]
        low = 0
        high = i - 1
        
        # Binary search posisi yang tepat
        while low <= high:
            mid = (low + high) // 2
            if ascending:
                if key < arr[mid]:
                    high = mid - 1
                else:
                    low = mid + 1
            else:
                if key > arr[mid]:
                    high = mid - 1
                else:
                    low = mid + 1
                    
        # Geser semua elemen ke kanan
        for j in range(i - 1, low - 1, -1):
            arr[j + 1] = arr[j]
        arr[low] = key
        
    return arr
```

```python
# --- MERGE SORT ---
def merge_sort(data, ascending=True):
    arr = data.copy()
    
    def _merge_sort(sub_arr):
        if len(sub_arr) <= 1:
            return sub_arr
            
        mid = len(sub_arr) // 2
        left = _merge_sort(sub_arr[:mid])
        right = _merge_sort(sub_arr[mid:])
        
        return _merge(left, right)
        
    def _merge(left, right):
        merged = []
        i = j = 0
        while i < len(left) and j < len(right):
            condition = left[i] <= right[j] if ascending else left[i] >= right[j]
            if condition:
                merged.append(left[i])
                i += 1
            else:
                merged.append(right[j])
                j += 1
                
        merged.extend(left[i:])
        merged.extend(right[j:])
        return merged

    return _merge_sort(arr)
```

```python
# --- QUICK SORT ---
def quick_sort(data, ascending=True, pivot_strategy="last"):
    arr = data.copy()
    
    def _quick_sort(low, high):
        if low < high:
            pi = _partition(low, high)
            _quick_sort(low, pi - 1)
            _quick_sort(pi + 1, high)
            
    def _partition(low, high):
        if pivot_strategy == "median":
            # Median-of-three pivot selection
            mid = (low + high) // 2
            candidates = [(arr[low], low), (arr[mid], mid), (arr[high], high)]
            candidates.sort(key=lambda x: x[0])
            median_idx = candidates[1][1]
            arr[median_idx], arr[high] = arr[high], arr[median_idx]
            
        pivot = arr[high]
        i = low - 1
        
        for j in range(low, high):
            condition = arr[j] <= pivot if ascending else arr[j] >= pivot
            if condition:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]
                
        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        return i + 1

    _quick_sort(0, len(arr) - 1)
    return arr
```

```python
# --- SHELL SORT ---
def shell_sort(data, ascending=True):
    arr = data.copy()
    n = len(arr)
    gap = n // 2
    
    while gap > 0:
        for i in range(gap, n):
            temp = arr[i]
            j = i
            # Perbandingan elemen berjarak 'gap'
            while j >= gap:
                condition = arr[j - gap] > temp if ascending else arr[j - gap] < temp
                if condition:
                    arr[j] = arr[j - gap]
                    j -= gap
                else:
                    break
            arr[j] = temp
        gap //= 2
        
    return arr
```

```python
# --- NON-COMPARISON SORT (COUNTING & RADIX SORT) ---
def counting_sort_nonnegative(data, ascending=True):
    # Hanya berfungsi untuk data bulat non-negatif
    if not data:
        return data
    arr = data.copy()
    max_val = max(arr)
    count = [0] * (max_val + 1)
    
    for num in arr:
        count[num] += 1
        
    sorted_arr = []
    if ascending:
        for val, freq in enumerate(count):
            sorted_arr.extend([val] * freq)
    else:
        for val in range(len(count) - 1, -1, -1):
            sorted_arr.extend([val] * count[val])
            
    return sorted_arr

def radix_sort_nonnegative(data):
    # Radix sort (LSD) untuk bilangan bulat non-negatif
    if not data:
        return data
    arr = data.copy()
    max_val = max(arr)
    
    exp = 1
    while max_val // exp > 0:
        # Lakukan counting sort berdasarkan digit ke-exp
        n = len(arr)
        output = [0] * n
        count = [0] * 10
        
        for i in range(n):
            index = (arr[i] // exp) % 10
            count[index] += 1
            
        for i in range(1, 10):
            count[i] += count[i - 1]
            
        i = n - 1
        while i >= 0:
            index = (arr[i] // exp) % 10
            output[count[index] - 1] = arr[i]
            count[index] -= 1
            i -= 1
            
        for i in range(n):
            arr[i] = output[i]
            
        exp *= 10
        
    return arr
```

```python
# --- VERIFIKASI & KOMPARASI KINERJA SORTING ---
import time
import random

print("=== VERIFIKASI KEBENARAN ALGORITMA SORTING ===")
angka_tes = [64, 34, 25, 12, 22, 11, 90]
names_tes = ["Sari", "Ahmad", "Budi", "Rina", "Doni"]

print(f"Data Asli Angka : {angka_tes}")
print(f"Data Asli Nama  : {names_tes}")
print("-" * 50)

# Verifikasi Bubble, Selection, Insertion
res_bubble, _, _ = bubble_sort(angka_tes)
print(f"Bubble Sort (Asc)      : {res_bubble}")
res_select, _, _ = selection_sort(angka_tes)
print(f"Selection Sort (Asc)   : {res_select}")
res_insert, _, _ = insertion_sort(angka_tes)
print(f"Insertion Sort (Asc)   : {res_insert}")
print(f"Binary Insertion (Asc) : {binary_insertion_sort(angka_tes)}")

# Verifikasi Merge, Quick, Shell, Counting, Radix
print(f"Merge Sort (Asc)       : {merge_sort(angka_tes)}")
print(f"Quick Sort (Asc)       : {quick_sort(angka_tes, pivot_strategy='median')}")
print(f"Shell Sort (Asc)       : {shell_sort(angka_tes)}")
print(f"Counting Sort (Asc)    : {counting_sort_nonnegative(angka_tes)}")
print(f"Radix Sort (Asc)       : {radix_sort_nonnegative(angka_tes)}")

# Verifikasi Pengurutan Nama
print(f"Selection Sort Nama    : {selection_sort_names(names_tes)}")
print(f"Merge Sort Nama (Desc) : {merge_sort(names_tes, ascending=False)}")

print("\n=== BENCHMARK KINERJA SORTING (WAKTU EKSEKUSI) ===")
# Membuat array acak berukuran 1000 elemen
random.seed(42)
array_besar = [random.randint(0, 10000) for _ in range(1000)]

sorting_algorithms = {
    "Bubble Sort": lambda d: bubble_sort(d),
    "Selection Sort": lambda d: selection_sort(d),
    "Insertion Sort": lambda d: insertion_sort(d),
    "Binary Insertion": lambda d: binary_insertion_sort(d),
    "Merge Sort": lambda d: merge_sort(d),
    "Quick Sort (Median)": lambda d: quick_sort(d, pivot_strategy='median'),
    "Shell Sort": lambda d: shell_sort(d),
    "Counting Sort": lambda d: counting_sort_nonnegative(d),
    "Radix Sort": lambda d: radix_sort_nonnegative(d)
}

print(f"Menguji kecepatan pada 1000 integer acak:")
print(f"{'Algoritma':<25} | {'Waktu Eksekusi (detik)':<25}")
print("-" * 55)

for name, sort_func in sorting_algorithms.items():
    start_time = time.perf_counter()
    sort_func(array_besar)
    elapsed = time.perf_counter() - start_time
    print(f"{name:<25} | {elapsed:.6f} s")
```
