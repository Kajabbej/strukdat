# 5. Linked List

```python
# --- SINGLY LINKED LIST (CORE) ---
class SinglyNode:
    def __init__(self, data):
        self.data = data
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def add_at_head(self, data):
        new_node = SinglyNode(data)
        new_node.next = self.head
        self.head = new_node

    def append(self, data):
        new_node = SinglyNode(data)
        if self.is_empty():
            self.head = new_node
            return
        curr = self.head
        while curr.next is not None:
            curr = curr.next
        curr.next = new_node

    def get_size(self):
        count = 0
        curr = self.head
        while curr is not None:
            count += 1
            curr = curr.next
        return count

    def search(self, data):
        curr = self.head
        while curr is not None:
            if curr.data == data:
                return True
            curr = curr.next
        return False

    def display(self):
        curr = self.head
        nodes = []
        while curr is not None:
            nodes.append(str(curr.data))
            curr = curr.next
        print(" -> ".join(nodes) + " -> None")
```

```python
# --- SINGLY LINKED LIST (ADVANCED OPERATIONS) ---
class AdvancedSinglyLinkedList(SinglyLinkedList):
    def insert_after(self, target_data, new_data):
        curr = self.head
        while curr is not None:
            if curr.data == target_data:
                new_node = SinglyNode(new_data)
                new_node.next = curr.next
                curr.next = new_node
                return True
            curr = curr.next
        return False

    def insert_before(self, target_data, new_data):
        if self.is_empty():
            return False
        if self.head.data == target_data:
            self.add_at_head(new_data)
            return True
            
        prev = None
        curr = self.head
        while curr is not None:
            if curr.data == target_data:
                new_node = SinglyNode(new_data)
                new_node.next = curr
                prev.next = new_node
                return True
            prev = curr
            curr = curr.next
        return False

    def delete_value(self, data):
        if self.is_empty():
            return False
        if self.head.data == data:
            self.head = self.head.next
            return True
            
        prev = None
        curr = self.head
        while curr is not None:
            if curr.data == data:
                prev.next = curr.next
                return True
            prev = curr
            curr = curr.next
        return False

    def reverse(self):
        prev = None
        curr = self.head
        while curr is not None:
            next_node = curr.next
            curr.next = prev
            prev = curr
            curr = next_node
        self.head = prev
```

```python
# --- DOUBLY LINKED LIST (CORE) ---
class DoublyNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def is_empty(self):
        return self.head is None

    def add_at_head(self, data):
        new_node = DoublyNode(data)
        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

    def append(self, data):
        new_node = DoublyNode(data)
        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node

    def display_forward(self):
        curr = self.head
        nodes = []
        while curr is not None:
            nodes.append(str(curr.data))
            curr = curr.next
        print("None <- " + " <-> ".join(nodes) + " -> None")

    def display_backward(self):
        curr = self.tail
        nodes = []
        while curr is not None:
            nodes.append(str(curr.data))
            curr = curr.prev
        print("None <- " + " <-> ".join(nodes) + " -> None")
```

```python
# --- DOUBLY LINKED LIST (ADVANCED OPERATIONS) ---
class AdvancedDoublyLinkedList(DoublyLinkedList):
    def insert_after(self, target_data, new_data):
        curr = self.head
        while curr is not None:
            if curr.data == target_data:
                new_node = DoublyNode(new_data)
                new_node.next = curr.next
                new_node.prev = curr
                
                if curr.next is not None:
                    curr.next.prev = new_node
                else:
                    self.tail = new_node # update tail
                curr.next = new_node
                return True
            curr = curr.next
        return False

    def insert_before(self, target_data, new_data):
        curr = self.head
        while curr is not None:
            if curr.data == target_data:
                new_node = DoublyNode(new_data)
                new_node.next = curr
                new_node.prev = curr.prev
                
                if curr.prev is not None:
                    curr.prev.next = new_node
                else:
                    self.head = new_node # update head
                curr.prev = new_node
                return True
            curr = curr.next
        return False

    def delete_value(self, data):
        curr = self.head
        while curr is not None:
            if curr.data == data:
                if curr.prev is not None:
                    curr.prev.next = curr.next
                else:
                    self.head = curr.next
                    
                if curr.next is not None:
                    curr.next.prev = curr.prev
                else:
                    self.tail = curr.prev
                return True
            curr = curr.next
        return False

    def reverse(self):
        curr = self.head
        self.tail = curr
        prev_node = None
        while curr is not None:
            # Tukar next dan prev
            prev_node = curr.prev
            curr.prev = curr.next
            curr.next = prev_node
            curr = curr.prev # pindah ke node berikutnya (yang tadinya adalah next)
        if prev_node is not None:
            self.head = prev_node.prev
```

```python
# --- CIRCULAR LINKED LIST ---
class CircularLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def append(self, data):
        new_node = SinglyNode(data)
        if self.is_empty():
            self.head = new_node
            new_node.next = self.head
        else:
            curr = self.head
            while curr.next != self.head:
                curr = curr.next
            curr.next = new_node
            new_node.next = self.head

    def prepend(self, data):
        new_node = SinglyNode(data)
        if self.is_empty():
            self.head = new_node
            new_node.next = self.head
        else:
            curr = self.head
            while curr.next != self.head:
                curr = curr.next
            new_node.next = self.head
            curr.next = new_node
            self.head = new_node

    def delete_value(self, data):
        if self.is_empty():
            return False
            
        curr = self.head
        prev = None
        
        # Kasus 1: Penghapusan head node
        if curr.data == data:
            # Jika hanya ada satu node
            if curr.next == self.head:
                self.head = None
                return True
            # Jika ada node lain, cari node terakhir
            last = self.head
            while last.next != self.head:
                last = last.next
            self.head = curr.next
            last.next = self.head
            return True
            
        # Kasus 2: Elemen lain
        prev = self.head
        curr = self.head.next
        while curr != self.head:
            if curr.data == data:
                prev.next = curr.next
                return True
            prev = curr
            curr = curr.next
        return False

    def display(self):
        if self.is_empty():
            print("None (Circular)")
            return
        curr = self.head
        nodes = []
        while True:
            nodes.append(str(curr.data))
            curr = curr.next
            if curr == self.head:
                break
        print(" -> ".join(nodes) + " -> (HEAD)")
```

```python
# --- VERIFIKASI LINKED LIST SUITE ---
print("=== 1. VERIFIKASI SINGLY LINKED LIST ===")
sll = AdvancedSinglyLinkedList()
sll.add_at_head(45)
sll.add_at_head(34)
sll.add_at_head(70)
print("Kondisi Awal Singly LinkedList:")
sll.display()

print("Insert 99 SETELAH 34:")
sll.insert_after(34, 99)
sll.display()

print("Insert 11 SEBELUM 45:")
sll.insert_before(45, 11)
sll.display()

print("Ukuran LinkedList:", sll.get_size())
print("Cari kunci 99 :", sll.search(99))
print("Cari kunci 100:", sll.search(100))

print("Penghapusan nilai 34:")
sll.delete_value(34)
sll.display()

print("Membalikkan Linked List (Reverse):")
sll.reverse()
sll.display()

print("\n=== 2. VERIFIKASI DOUBLY LINKED LIST ===")
dll = AdvancedDoublyLinkedList()
dll.append(10)
dll.append(20)
dll.append(30)
print("Doubly LinkedList (Maju):")
dll.display_forward()
print("Doubly LinkedList (Mundur):")
dll.display_backward()

print("Insert 15 sebelum 20:")
dll.insert_before(20, 15)
dll.display_forward()

print("Insert 25 setelah 20:")
dll.insert_after(20, 25)
dll.display_forward()

print("Hapus nilai 15:")
dll.delete_value(15)
dll.display_forward()

print("Membalikkan Doubly LinkedList:")
dll.reverse()
dll.display_forward()

print("\n=== 3. VERIFIKASI CIRCULAR LINKED LIST ===")
cll = CircularLinkedList()
cll.append(1)
cll.append(2)
cll.append(3)
print("Circular LinkedList:")
cll.display()

print("Prepend nilai 0:")
cll.prepend(0)
cll.display()

print("Hapus nilai 2:")
cll.delete_value(2)
cll.display()

print("Hapus head (nilai 0):")
cll.delete_value(0)
cll.display()
```
