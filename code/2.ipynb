# 2. Infix, Prefix, & Postfix

```python
# --- STACK HELPER UNTUK EKSPRESI ---
class Stack:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.isEmpty():
            return None
        return self.items.pop()

    def peek(self):
        if self.isEmpty():
            return None
        return self.items[-1]

    def size(self):
        return len(self.items)
```

```python
# --- KONVERSI INFIX KE POSTFIX ---
def infix_to_postfix(ekspresi):
    presedensi = {'+': 1, '-': 1, '*': 2, '/': 2, '^': 3}
    stack = Stack()
    output = []
    
    # Hapus spasi
    tokens = ekspresi.replace(" ", "")
    
    for token in tokens:
        if token.isalnum():
            output.append(token)
        elif token == '(':
            stack.push(token)
        elif token == ')':
            while not stack.isEmpty() and stack.peek() != '(':
                output.append(stack.pop())
            stack.pop() # Buang '('
        else: # Operator
            while (not stack.isEmpty() and stack.peek() != '(' and
                   presedensi.get(stack.peek(), 0) >= presedensi.get(token, 0)):
                output.append(stack.pop())
            stack.push(token)
            
    while not stack.isEmpty():
        output.append(stack.pop())
        
    return " ".join(output)
```

```python
# --- KONVERSI INFIX KE PREFIX ---
def infix_to_prefix(ekspresi):
    # Balik ekspresi dan sesuaikan kurung
    dibalik = ekspresi.replace(" ", "")[::-1]
    temp = []
    for char in dibalik:
        if char == '(':
            temp.append(')')
        elif char == ')':
            temp.append('(')
        else:
            temp.append(char)
    ekspresi_balik = "".join(temp)
    
    # Konversi ke postfix
    postfix = infix_to_postfix(ekspresi_balik)
    
    # Balik hasil postfix untuk mendapatkan prefix
    return " ".join(postfix.split()[::-1])
```

```python
# --- EVALUASI EKSPRESI POSTFIX ---
def evaluasi_postfix(ekspresi):
    stack = Stack()
    tokens = ekspresi.split()
    
    for token in tokens:
        if token.isalnum():
            # Cek jika token numerik desimal atau bulat
            if token.isdigit():
                stack.push(int(token))
            else:
                try:
                    stack.push(float(token))
                except ValueError:
                    # Jika berupa variabel, beri nilai dummy atau error
                    raise ValueError(f"Variabel '{token}' tidak bisa dievaluasi tanpa nilai!")
        else:
            operand2 = stack.pop()
            operand1 = stack.pop()
            
            if token == '+':
                res = operand1 + operand2
            elif token == '-':
                res = operand1 - operand2
            elif token == '*':
                res = operand1 * operand2
            elif token == '/':
                res = operand1 / operand2
            elif token == '^':
                res = operand1 ** operand2
            stack.push(res)
            
    return stack.pop()
```

```python
# --- EVALUASI EKSPRESI PREFIX ---
def evaluasi_prefix(ekspresi):
    stack = Stack()
    # Evaluasi dari kanan ke kiri
    tokens = ekspresi.split()[::-1]
    
    for token in tokens:
        if token.isalnum():
            if token.isdigit():
                stack.push(int(token))
            else:
                try:
                    stack.push(float(token))
                except ValueError:
                    raise ValueError(f"Variabel '{token}' tidak bisa dievaluasi tanpa nilai!")
        else:
            operand1 = stack.pop()
            operand2 = stack.pop()
            
            if token == '+':
                res = operand1 + operand2
            elif token == '-':
                res = operand1 - operand2
            elif token == '*':
                res = operand1 * operand2
            elif token == '/':
                res = operand1 / operand2
            elif token == '^':
                res = operand1 ** operand2
            stack.push(res)
            
    return stack.pop()
```

```python
# --- KONVERSI POSTFIX KE INFIX ---
def postfix_to_infix(ekspresi):
    stack = Stack()
    tokens = ekspresi.split()
    
    for token in tokens:
        if token.isalnum():
            stack.push(token)
        else:
            op2 = stack.pop()
            op1 = stack.pop()
            stack.push(f"({op1} {token} {op2})")
            
    return stack.pop()
```

```python
# --- KONVERSI PREFIX KE INFIX ---
def prefix_to_infix(ekspresi):
    stack = Stack()
    # Proses dari kanan ke kiri
    tokens = ekspresi.split()[::-1]
    
    for token in tokens:
        if token.isalnum():
            stack.push(token)
        else:
            op1 = stack.pop()
            op2 = stack.pop()
            stack.push(f"({op1} {token} {op2})")
            
    return stack.pop()
```

```python
# --- VERIFIKASI & DEMONSTRASI SELURUH OPERASI ---
print("=== VERIFIKASI KONVERSI & EVALUASI EKSPRESI MATEMATIKA ===")
print("----------------------------------------------------------")

infix_examples = [
    "A + B * C",
    "(A + B) * C",
    "A * (B + C)",
    "(A - B) / (C + D)",
    "A ^ B * C + D"
]

print("1. Konversi Infix ke Postfix & Prefix:")
for expr in infix_examples:
    postfix = infix_to_postfix(expr)
    prefix = infix_to_prefix(expr)
    print(f"Infix   : {expr}")
    print(f"Postfix : {postfix}")
    print(f"Prefix  : {prefix}")
    print("-" * 40)

print("\n2. Evaluasi Numerik Ekspresi Postfix & Prefix:")
post_expr1 = "5 3 2 * +"  # 5 + (3 * 2) = 11
pre_expr1 = "+ 5 * 3 2"
post_expr2 = "2 3 + 2 ^ 5 -"  # (2 + 3) ^ 2 - 5 = 20
pre_expr2 = "- ^ + 2 3 2 5"

print(f"Postfix '{post_expr1}' -> Hasil: {evaluasi_postfix(post_expr1)}")
print(f"Prefix  '{pre_expr1}' -> Hasil: {evaluasi_prefix(pre_expr1)}")
print(f"Postfix '{post_expr2}' -> Hasil: {evaluasi_postfix(post_expr2)}")
print(f"Prefix  '{pre_expr2}' -> Hasil: {evaluasi_prefix(pre_expr2)}")

print("\n3. Konversi Balik (Postfix/Prefix ke Infix):")
post_sym = "A B + C *"
pre_sym = "* + A B C"
print(f"Postfix '{post_sym}' -> Infix: {postfix_to_infix(post_sym)}")
print(f"Prefix  '{pre_sym}' -> Infix: {prefix_to_infix(pre_sym)}")
```
