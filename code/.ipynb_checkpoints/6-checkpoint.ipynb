### 6. Tree

```python
class BSTNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

    # ========================
    # INSERT
    # ========================
    def insert(self, value):
        if value == self.value:
            return

        if value < self.value:
            if self.left is None:
                self.left = BSTNode(value)
            else:
                self.left.insert(value)
        else:
            if self.right is None:
                self.right = BSTNode(value)
            else:
                self.right.insert(value)

    # ========================
    # SEARCH
    # ========================
    def search(self, value):
        if value == self.value:
            return True
        elif value < self.value and self.left:
            return self.left.search(value)
        elif value > self.value and self.right:
            return self.right.search(value)
        return False

    # ========================
    # FIND MIN
    # ========================
    def find_min(self):
        current = self
        while current.left:
            current = current.left
        return current

    # ========================
    # DELETE
    # ========================
    def delete(self, value):
        if value < self.value:
            if self.left:
                self.left = self.left.delete(value)

        elif value > self.value:
            if self.right:
                self.right = self.right.delete(value)

        else:
            # tidak punya anak
            if self.left is None and self.right is None:
                return None

            # satu anak
            if self.left is None:
                return self.right
            if self.right is None:
                return self.left

            # dua anak
            min_larger_node = self.right.find_min()
            self.value = min_larger_node.value
            self.right = self.right.delete(min_larger_node.value)

        return self

    # ========================
    # TRAVERSAL
    # ========================
    def inorder(self):
        if self.left:
            self.left.inorder()
        print(self.value, end=" ")
        if self.right:
            self.right.inorder()

    # ========================
    # VISUALISASI
    # ========================
    def display(self, level=0, prefix="Root: "):
        print(" " * (level * 4) + prefix + str(self.value))

        if self.left:
            self.left.display(level + 1, "L--- ")
        if self.right:
            self.right.display(level + 1, "R--- ")
```

```python
# buat root
root = BSTNode(50)

# insert data
data = [30, 70, 20, 40, 60, 80]
for d in data:
    root.insert(d)

# tampilkan tree
root.display()

# traversal
print("\nInorder:")
root.inorder()

# search
print("\n\nSearch 40:", root.search(40))

# delete
root = root.delete(50)

print("\n\nSetelah delete root:")
root.display()
```
